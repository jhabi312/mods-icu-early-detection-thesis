```python
import pandas as pd
import matplotlib.pyplot as plt

# Load sample ICU dataset
df = pd.read_csv('sample_data.csv')

# Basic exploration
print("Dataset shape:", df.shape)
print("\nSummary statistics:")
print(df.describe())

# Check for missing values
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Plot MAP trend over time
plt.figure(figsize=(10, 5))
plt.plot(df['hour'], df['MAP_mmHg'], marker='o', color='steelblue', label='MAP (mmHg)')
plt.axhline(y=65, color='red', linestyle='--', label='Critical threshold (65 mmHg)')
plt.title('Mean Arterial Pressure Trend — ICU Post-Operative Patient')
plt.xlabel('Hour Post-Surgery')
plt.ylabel('MAP (mmHg)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Plot lactic acid trend
plt.figure(figsize=(10, 5))
plt.plot(df['hour'], df['lactic_acid_mmol'], marker='s', color='tomato', label='Lactic Acid (mmol/L)')
plt.axhline(y=2.0, color='darkred', linestyle='--', label='Elevated threshold (2.0 mmol/L)')
plt.title('Lactic Acid Trend — ICU Post-Operative Patient')
plt.xlabel('Hour Post-Surgery')
plt.ylabel('Lactic Acid (mmol/L)')
plt.legend()
plt.tight_layout()
plt.show()

print("\nAnalysis complete.")
```
